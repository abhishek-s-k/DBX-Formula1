```python
source_file = f"{landing_folder_path}/{v_batch_id}/constructors.json"
table_name = f"{catalog_name}.{bronze_schema}.constructors"
```

### Step 1 - Read the JSON file using DataFrameReader API

```python
constructors_schema = "constructorId STRING, name STRING, nationality STRING, url STRING"
```

```python
constructors_df = (
    spark.read.format('json')
    .schema(constructors_schema)
    .option('mode','FAILFAST')
    .load(source_file)
) 
```


### Step 2 - Add Metadata Columns
- Source File
- Ingestion Timestamp

```python
constructors_final_df = add_ingestion_metadata(constructors_df)
```

### Step 3 - Write to Bronze Delta Table

```python
write_to_bronze(input_df = constructors_final_df, table_name = table_name, batch_id = v_batch_id)
```
