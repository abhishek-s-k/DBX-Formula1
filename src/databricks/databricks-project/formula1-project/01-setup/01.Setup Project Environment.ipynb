# Set-up the project environment for Formula-1 Project

1. Create External Location
2. Create Catalog 'Formula'
3. Create Schemas 'landing', 'bronze', 'silver', 'gold'
4. create volume 'Files' in the landing schema

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_project_ext_dl_formula1
URL 'abfss://formula1@databricksprojectextdl.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `databricks-project-sc`)
COMMENT 'created external location for formula-1 project container'
```

```sql
%fs ls 'abfss://formula1@databricksprojectextdl.dfs.core.windows.net/landing'
```

## Create Catalog 'Formula1'

```sql
SHOW CATALOGS;
```

```sql
CREATE CATALOG IF NOT EXISTS formula1
MANAGED LOCATION 'abfss://formula1@databricksprojectextdl.dfs.core.windows.net/'
COMMENT 'created catalog for formula-1 project';
```

## Create Schemas

```sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;
CREATE SCHEMA IF NOT EXISTS formula1.bronze
    MANAGED LOCATION 'abfss://formula1@databricksprojectextdl.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1.silver
    MANAGED LOCATION 'abfss://formula1@databricksprojectextdl.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS formula1.gold
    MANAGED LOCATION 'abfss://formula1@databricksprojectextdl.dfs.core.windows.net/gold';
```

```sql
USE CATALOG formula1;
```

```sql
SELECT current_catalog();
```

```sql
SHOW SCHEMAS;
```

```sql
CREATE EXTERNAL VOLUME formula1.landing.files
LOCATION 'abfss://formula1@databricksprojectextdl.dfs.core.windows.net/landing'
COMMENT 'landing files volume for formula-1 project';
```

```sql
%fs ls /Volumes/formula1/landing/files
```
