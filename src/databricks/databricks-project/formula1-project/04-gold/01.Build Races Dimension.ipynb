## Build Races Dimension

1. Read silver races table
2. Read silver circuits table
3. Join the data from races with circuits using circuit_id
4. Select the required columns
    - races.season
    - races.round
    - races.race_name
    - races.race_date
    - circuits.circuit_name
    - circuits.locality
    - circuits.country
5. Write the transformed data to gold dim_races

```python
%run ../00-Common/01.environment-config
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
target_table = f"{catalog_name}.{gold_schema}.dim_races"
```

### Step 1 -  Read silver races table


```python
races_df = spark.table(f"{catalog_name}.{silver_schema}.races")
```


### Step 2 - Read silver circuits table

```python
circuits_df = spark.table(f"{catalog_name}.{silver_schema}.circuits")
```


### Step 3 - Join the data from races with circuits using circuit_id


```python
dim_races_df = races_df.join(circuits_df, races_df.circuit_id == circuits_df.circuit_id, "inner")
```

### Step 4 - Select the required columns
-      races.season
-      races.round
-      races.race_name
-      races.race_date
-      circuits.circuit_name
-      circuits.locality
-      circuits.country

```python
dim_races_df = dim_races_df.select(
    races_df.season,
    races_df.round,
    races_df.race_name,
    races_df.race_date,
    circuits_df.circuit_name,
    circuits_df.locality,
    circuits_df.country
)
```

### Step 5 - Write the transformed data to gold dim_races

```python
(
    dim_races_df
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(target_table)
)
```
