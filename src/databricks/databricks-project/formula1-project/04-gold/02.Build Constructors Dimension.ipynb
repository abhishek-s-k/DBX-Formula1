## Build Constructors Dimension

1. Read silver constructors table
2. Read gold ref_nationality_region table
3. Join the data from constructors with ref_nationality_region using nationality
4. Select the required columns
    - constructors.constructors_id
    - constructors.constructors_name
    - constructors.nationality
    - ref_nationality_region.region
5. Write the transformed data to gold dim_constructors table

```python
%run ../00-Common/01.environment-config
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
target_table = f"{catalog_name}.{gold_schema}.dim_constructors"
```

### Step 1 -  Read silver constructors table


```python
constructors_df = spark.table(f"{catalog_name}.{silver_schema}.constructors")
```


### Step 2 - Read gold ref_nationality_region table

```python
ref_nationality_region_df = spark.table(f"{catalog_name}.{gold_schema}.ref_nationality_region")
```


### Step 3 - Join the data from constructors with ref_nationality_region using nationality


```python
dim_constructors_df = constructors_df.join(ref_nationality_region_df, constructors_df.nationality == ref_nationality_region_df.nationality, "left")
```

### Step 4. Select the required columns
-     constructors.constructors_id
-     constructors.constructors_name
-     constructors.nationality
-     ref_nationality_region.region

```python
dim_constructors_df = dim_constructors_df.select(
    constructors_df.constructor_id,
    constructors_df.constructor_name,
    constructors_df.nationality,
    ref_nationality_region_df.region.alias("nationality_region")
)
```

### Step 5 - Write the transformed data to gold dim_constructors table

```python
(
    dim_constructors_df
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(target_table)
)
```
